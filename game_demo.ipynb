```python
def battle(merged_df):
    text_input = widgets.Text(
        placeholder = 'choice',
        description = 'choose your pokemon',
        layout = widgets.Layout(width='300px')
    )
    button = widgets.Button(
        description = 'ready for battle',
        button_style = 'danger'
    )
    output = widgets.Output()

    def on_battle_click(b):
        with output:
            clear_output()
            user_name = text_input.value.strip().capitalize()
            
            #match input name with row in our df
            match = merged_df[merged_df['name'].str == 'user_name']

            #if they made up pokemon
            if match.empty:
                print("that's not a known pokemon, try another one pls")
                return

            my_pokemon = match
            opp_pokemon = merged_df.sample().iloc[0]

            if my_pokemon.hp > opp_pokemon.hp:
                print(f"{my_pokemon["name"]} wins!")
                get_image_from_url(my_pokemon.image_url)
            elif my_pokemon.hp < opp_pokemon.hp:
                print(f"{opp_pokemon["name"]} wins!")
                get_image_from_url(opp_pokemon.image_url)
            else:
                 print(f"it's a tie...")
    button.on_click(on_battle_click)
    display(widgets.VBox([text_input, button, output]))
```

```python
battle(merged_df)
```
